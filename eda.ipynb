Function for EDA across audio domains

```python
def eda_report(dataset, name):
    print(f"---{name}---")
    print(f"Total samples: {len(dataset)}")

    labels = [entry["label"] for entry in dataset.data]
    pos = sum(labels)
    neg = len(labels) - pos
    print(f"Positive samples: {pos} ({pos/len(labels):.1%})\nNegative samples: {neg} ({neg/len(labels):.1%})")

    print(f"Unique label values seen: {set(labels)}")

    durations = []
    sampling_rates = set()
    n_errors = 0

    for entry in dataset.data:
        try:
            waveform, sr = dataset.read_audio(entry["audio"])
            durations.append(len(waveform)/sr)
            sampling_rates.add(sr)
        except Exception as e:
            n_errors +=1
            print(f"Failed: {entry['audio'] if not isinstance(entry['audio'], bytes) else '[bytes]'} — {e}")

    print(f"Duration (sampled): min = {min(durations):.2f}s max = {max(durations):.2f}s mean = {np.mean(durations):.2f}s")
    durations_sorted = sorted(durations)
    p95 = durations_sorted[int(len(durations_sorted) * 0.95)]
    print(f"95th percentile duration: {p95:.2f}s")

    
    print(f"Sampling rates seen: {sampling_rates}")
    print(f"Unreadable files (of sample): {n_errors}")
    print()
```

Speech dataset (CAMEO)

```python
speech_ds = SpeechDataset()
eda_report(speech_ds, "CAMEO")

n_ids = len(set(entry["group"] for entry in speech_ds.data))
print(f"Unique speakers: {n_ids}")
```

Music dataset (PMEmo)

```python
music_ds = MusicDataset()
eda_report(music_ds, "PMEmo")
n_ids_music = len(set(entry["group"] for entry in music_ds.data))
print(f"Unique artists: {n_ids_music}")
```

Animal vocalizations dataset

```python
animals_ds = AnimalVocDataset()
eda_report(animals_ds, "Animal vocalizations dataset")
n_ids_animals = len(set(entry["group"] for entry in animals_ds.data))
print(f"Unique animal IDs: {n_ids_animals}")
```

Soundscapes dataset (Emo-Soundscapes)

```python
soundscapes_ds = SoundscapesDataset()
eda_report(soundscapes_ds, "Emo-Soundscapes")
```
